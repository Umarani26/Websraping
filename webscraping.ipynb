```python
from bs4 import BeautifulSoup
import requests
```

```python
Download the contents of the webpage
```

```python
#this url contains the data you need to scrape
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"
```

```python
# get the contents of the webpage in text format and store in a variable called data
data=requests.get(url).text
```

```python
Create a soup object using the class BeautifulSoup
```

```python
soup=BeautifulSoup(data, 'html.parser')
```

```python
Scrape  all images
```

```python
for link in soup.find_all('a'):
    print(link.get("href"))
```

```python
for image in soup.find_all('img'):
    print(image.get('src'))
```

```python
find a html table in the web page
```

```python
table=soup.find('table')
```

```python
Get all rows from the table
```

```python
scrapped_data=[]
for row in table.find_all('tr'):
    col=row.find_all('td')
    language=col[1].getText()
    average_annual_salary=col[3].getText()
    scrapped_data.append([language, average_annual_salary])
    print("{:10}| {}".format(language,average_annual_salary))
print("scrapped_data",scrapped_data)    
```

```python
import pandas as pd
```

```python
df=pd.DataFrame(scrapped_data, columns=["Language", "Average Annual Salary"])
```

```python
df.to_csv("popular-languages.csv", index=False, encoding="utf-8")
```
